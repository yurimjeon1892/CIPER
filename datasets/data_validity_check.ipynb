```python
from PIL import Image
import numpy as np
import os
```

```python
data_root = "/storage2/public/kitti"
train_pt_list = "./splits/kitti/train_files.txt"
val_same_pt_list = "./splits/kitti/test1_files.txt"
val_cross_pt_list = "./splits/kitti/test2_files.txt"
```

```python
def make_sample_list_kitti(data_root, pt_list):
    with open(pt_list, "r") as f:
        file_name_list = f.readlines()

    sample_list = []
    for file_ in file_name_list:
        file_ = file_[:-1]
        sample_list.append(file_)
    print(len(sample_list))
    
    cnt_ground, cnt_aerial = 0,0
        
    for sample in sample_list:
        
        file_name = sample.split(" ")[0]
        # day_dir = file_name[:10]
        drive_dir = file_name[:38]
        image_no = file_name[38:]

        # =================== read ground image ===================================
        left_img_name = os.path.join(
            data_root, "raw", drive_dir, "image_02/data", image_no.lower()
        )
        if os.path.isfile(left_img_name):
            cnt_ground += 1

        # =================== read satellite map ==================================
        arl_img_name = os.path.join(data_root, "satellite", file_name)
        if os.path.isfile(arl_img_name):
            cnt_aerial += 1
    print(pt_list, cnt_ground, cnt_aerial)
```

```python
make_sample_list_kitti(data_root, train_pt_list)
make_sample_list_kitti(data_root, val_same_pt_list)
make_sample_list_kitti(data_root, val_cross_pt_list)
```
